### 1. Contexto del problema trabajado en la actividad de exploración

El objetivo de nuestro proyecto es diseñar una base de datos relacional que permita gestionar el funcionamiento de una clínica veterinaria con servicios y tamaño dinámico que se enfoca en animales domésticos y todo el proceso requerido para la atención en salud animal. Para abordar correctamente el modelado, es fundamental comprender los siguientes conceptos:

* **Historia Clínica Electrónica Veterinaria (HCEV):** Es el eje central del sistema, con una estructura jerárquica donde un Propietario está vinculado a uno o varios Pacientes (mascotas). La base de datos debe garantizar la integridad de esta relación.
* **Gestión de Catálogos Dinámicos:** Se requiere parametrización mediante tablas maestras para que los administradores puedan crear, modificar o desactivar servicios sin alterar el esquema de la base de datos, evitando codificar los servicios de forma estática.
* **Trazabilidad del Flujo de Atención:** La base de datos debe rastrear el estado del paciente en cada etapa del ciclo: Agendamiento, Admisión, Atención Médica, Servicios Auxiliares, Facturación y Seguimiento.
* **Control de Inventario Farmacológico:** Existe una relación crítica entre los procedimientos realizados y el descargo automático de medicamentos o insumos médicos del inventario, gestionando unidades de medida precisas.

### 2. Consulta de tendencias actuales en el área del proyecto

Las tendencias que dictan cómo se construyen estas bases de datos hoy en día incluyen:

* **Arquitecturas Cloud-Native (SaaS) para Escalabilidad Dinámica:** Utilizan bases de datos en la nube que escalan automáticamente según el volumen de datos para cumplir con el requerimiento de tamaño dinámico.
* **Interoperabilidad e Integración con IoT (Internet of Things):** Las bases de datos se diseñan para recibir información externa, como integraciones con analizadores de sangre o collares inteligentes.
* **Telemedicina y Tele-Triage Veterinario:** Los sistemas incorporan módulos de consultas virtuales, lo que requiere tipificar los servicios en la base de datos y almacenar enlaces digitales.
* **Uso de Modelos de Datos Híbridos (Relacional + JSON):** Hay una tendencia a utilizar campos estructurados tipo JSON dentro de bases de datos relacionales para almacenar formularios médicos altamente dinámicos, manteniendo el núcleo financiero estrictamente relacional.

### 3. Consulta de herramientas o sistemas similares con su análisis de funcionalidades

Analizamos dos plataformas líderes en la gestión veterinaria para entender las soluciones de la industria:

**A. Provet Cloud (por Nordhealth)**
* **Descripción:** Software de gestión basado en la nube, diseñado para ser altamente personalizable.
* **Manejo del tamaño dinámico:** Su arquitectura permite el uso por clínicas independientes y grandes universidades, evidenciando la necesidad de contemplar roles y sucursales en nuestra base.
* **Servicios dinámicos:** Permite crear paquetes donde un procedimiento agrupa automáticamente insumos, tiempo y equipos.
* **Oportunidad de mejora:** Su curva de aprendizaje es alta; nuestro diseño buscará un equilibrio entre flexibilidad y simplicidad.

**B. eVetPractice (por Covetrus)**
* **Descripción:** Sistema en la nube enfocado en la usabilidad y la integración de facturación clínica.
* **Flujo de atención:** Utiliza una pizarra electrónica que muestra la etapa clínica de cada mascota, sugiriendo que nuestra base debe incluir un atributo de estado de flujo de trabajo.
* **Integración financiera:** Relaciona el historial clínico con el carrito de cobro, insertando servicios automáticamente en la pre-factura, lo que requiere un diseño relacional fuerte.
